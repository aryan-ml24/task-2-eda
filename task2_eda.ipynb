# Task 2 — Exploratory Data Analysis (EDA)

**Dataset:** Tips dataset

**Objective:** Understand data using statistics and visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

```python
df = pd.read_csv('data/tips.csv')
df.head()
```

## 1. Dataset overview

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 2. Descriptive statistics

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
summary = df[numeric_cols].describe().T
summary['median'] = df[numeric_cols].median()
summary['skewness'] = df[numeric_cols].skew()
summary.round(2)
```

## 3. Histograms

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 8))
for ax, col in zip(axes.flat, numeric_cols):
    ax.hist(df[col], bins=20)
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Frequency')
plt.tight_layout()
```

## 4. Boxplots

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for ax, col in zip(axes.flat, numeric_cols):
    sns.boxplot(x=df[col], ax=ax)
    ax.set_title(f'Boxplot of {col}')
plt.tight_layout()
```

## 5. Correlation matrix

```python
corr = df[numeric_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', square=True)
plt.title('Correlation Matrix')
plt.show()
```

## 6. Feature relationships

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time', style='sex', alpha=0.75)
plt.title('Tip vs Total Bill')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(data=df, x='day', y='total_bill', ax=axes[0])
axes[0].set_title('Total Bill by Day')
sns.boxplot(data=df, x='time', y='tip', hue='sex', ax=axes[1])
axes[1].set_title('Tip by Meal Time and Sex')
plt.tight_layout()
```

## 7. Findings

- Total bill vs tip correlation: **0.68**.
- Total bill vs party size correlation: **0.60**.
- Larger bills generally correspond to larger absolute tips.
- Boxplots highlight spread and potential outliers.
- Correlation is association, not causation.

## 8. Interview Questions — Short Answers

1. **EDA purpose:** understand data quality, distributions, relationships, and anomalies before modeling.
2. **Boxplots:** summarize median, quartiles, spread, and potential outliers.
3. **Correlation:** strength/direction of linear association; not causation.
4. **Skewness:** inspect histograms/boxplots and calculate `.skew()`.
5. **Multicollinearity:** strong dependence among predictors; can affect regression coefficient stability.
6. **Tools:** Pandas, NumPy, Matplotlib, Seaborn, Plotly.
7. **EDA problem example:** missing values or extreme observations discovered before model training.
8. **Visualization in ML:** helps understand distributions, relationships, anomalies, and feature behavior.
